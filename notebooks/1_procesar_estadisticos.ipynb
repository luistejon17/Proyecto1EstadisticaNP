```python
import numpy as np
import pandas as pd
import os
import sys

sys.path.append("..")
import metodos_estadisticos as tests

ruta_data = "../data"
ruta_resultados = "../resultados"

resultados = []
```

```python
for escenario in range(1, 14):
    
    print(escenario)
    data = np.load(f"{ruta_data}/datos_escenario_{escenario}.npz")

    xs = data["x"]
    ys = data["y"]

    for i, (x, y) in enumerate(zip(xs, ys)):

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "t_student",
            "resultado": int(tests.test_t_student(x, y))
        })

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "wilcoxon",
            "resultado": int(tests.test_wilcoxon(x, y))
        })

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "hodges_lehmann",
            "resultado": int(tests.test_hodges_lehmann(x, y))
        })

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "perm_media",
            "resultado": int(tests.test_permutacion(x, y, "media"))
        })

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "perm_mediana",
            "resultado": int(tests.test_permutacion(x, y, "mediana"))
        })

        resultados.append({
            "escenario": escenario,
            "muestra": i,
            "metodo": "perm_trim",
            "resultado": int(tests.test_permutacion(x, y, "trim"))
        })


df = pd.DataFrame(resultados)

df.to_csv(f"{ruta_resultados}/resultados_tests.csv", index=False)
#df.to_parquet(f"{ruta_resultados}/resultados_tests.parquet", index=False)
```
