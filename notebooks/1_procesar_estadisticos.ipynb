```python
import numpy as np
import pandas as pd
import sys
import time

```

```python
sys.path.append("..")
import metodos_estadisticos as tests

ruta_data = "../data"
ruta_resultados = "../resultados"

resultados = []


for escenario in range(1, 2):
    print("evaluando procedimientos en escenario ", escenario)
    data = np.load(f"{ruta_data}/datos_escenario_{escenario}.npz")

    xs = data["x"]
    ys = data["y"]

    for i, (x, y) in enumerate(zip(xs, ys)):

        # t_student
        t0 = time.time()
        res = int(tests.test_t_student(x, y))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "t_student", "resultado": res, "tiempo": dt})

        # wilcoxon
        t0 = time.time()
        res = int(tests.test_wilcoxon(x, y))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "wilcoxon", "resultado": res, "tiempo": dt})

        # hodges_lehmann
        t0 = time.time()
        res = int(tests.test_hodges_lehmann(x, y))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "hodges_lehmann", "resultado": res, "tiempo": dt})

        #perm_media
        t0 = time.time()
        res = int(tests.test_permutacion(x, y, "media"))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "perm_media", "resultado": res, "tiempo": dt})

        # perm_mediana
        t0 = time.time()
        res = int(tests.test_permutacion(x, y, "mediana"))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "perm_mediana", "resultado": res, "tiempo": dt})

        # perm_trim
        t0 = time.time()
        res = int(tests.test_permutacion(x, y, "trim"))
        dt = time.time() - t0
        resultados.append({"escenario": escenario, "muestra": i, "metodo": "perm_trim", "resultado": res, "tiempo": dt})


df = pd.DataFrame(resultados)
df.to_csv(f"{ruta_resultados}/resultados_tests.csv", index=False)
```
