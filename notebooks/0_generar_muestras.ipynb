```python
import numpy as np
import os
import sys

sys.path.append("..")
import distribuciones as dist


ruta_data = os.path.join('..', 'data')

n_muestras = 500
```

```python
# Se crean los datos por cada posible índice y se almacenan en la carpeta data



for i in range(1, 14):

    xs = []
    ys = []

    for _ in range(n_muestras):
        x, y = dist.generar_muestras(escenario=i)
        xs.append(x)
        ys.append(y)

    xs = np.array(xs)
    ys = np.array(ys)

    np.savez_compressed(os.path.join(ruta_data,f"datos_escenario_{i}.npz"), x=xs, y=ys)
```
